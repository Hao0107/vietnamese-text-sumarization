```python
json_dataPath = '../data/summarized_2396_Docs.json'
outputDataPath = '../data/summarized_2396_Docs.xlsx'
```

```python
import pandas as pd
import json
import re

# Đọc dữ liệu
with open(json_dataPath, 'r', encoding='utf-8') as f:
    data = [json.loads(line) for line in f]

df = pd.DataFrame(data)

# Hàm loại bỏ ký tự không hợp lệ cho Excel
def remove_illegal_chars(val):
    if isinstance(val, str):
        # Loại bỏ các ký tự có mã ASCII từ 0-31 (ngoại trừ xuống dòng \n, tab \t, về đầu dòng \r)
        return re.sub(r'[\x00-\x08\x0b\x0c\x0e-\x1f]', '', val)
    return val

# Áp dụng làm sạch cho toàn bộ DataFrame
df = df.map(remove_illegal_chars)

df.to_excel(outputDataPath, index=False)

print(f"Xong! File {outputDataPath} đã được tạo thành công.")
```
