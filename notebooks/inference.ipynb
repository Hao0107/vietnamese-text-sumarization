```python
import evaluate
import json 
```

```python
test_datapath = "../data/processed/df_processed2603.jsonl"
```

```python
def run_eval(data_path=""):
    summarizer = Summarizer(model_path=model_path)
    rouge = evaluate.load("rouge")
    
    test_data = []
    with open(test_datapath, "r", encoding="utf-8") as f:
        for line_num, line in enumerate(f, 1):
            if not line.strip(): continue
            try:
                test_data.append(json.loads(line))
            except json.JSONDecodeError as e:
                print(f"Lỗi tại dòng {line_num}: {e}")

    refs = [d['processed_summary'] for d in test_data[100::-1]]
    preds = []

    print(f"Đang đánh giá trên {len(refs)} mẫu...")
    for i, d in enumerate(test_data[:100]):
        p = summarizer.summarize(d['processed_content'])
        preds.append(p)
        if i % 10 == 0: print(f"Đã xong {i} bài...")

    results = rouge.compute(predictions=preds, references=refs)
    print("\n--- KẾT QUẢ ĐÁNH GIÁ (ROUGE SCORE) ---")
    for k, v in results.items():
        print(f"{k}: {v*100:.2f}")
```

```python
run_eval(data_path=test_datapath)
```
